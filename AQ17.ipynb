```python
def find_highest_frequency(text):
    
    clean_text = [char.lower() for char in text if char.isalpha()]
    

    frequency_map = {}
    

    for char in clean_text:
        if char in frequency_map:
            frequency_map[char] += 1
        else:
            frequency_map[char] = 1
 
    max_char = ''
    max_count = 0
    
    for char, count in frequency_map.items():
        if count > max_count:
            max_char = char
            max_count = count
            
    return max_char, max_count

input_str = "Pratibha Pawar"

char, count = find_highest_frequency(input_str)

print(f"Input = {input_str}")
print(f"Output = {char} Occurred {count} times")
```
