```python
31. You have two tuples’ names = ('Alice', 'Bob') and ages = (25, 30). Write code to combine these
a. tuples into a list of tuples like [(Alice, 25), (Bob, 30)].
Note: use zip method
```

```python
names = ('Alice', 'Bob')
ages = (25, 30)

combined_list = list(zip(names, ages))

print(combined_list)
```
