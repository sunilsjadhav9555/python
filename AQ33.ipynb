```python
34. You have a dictionary score = {'Alice': 90, 'Bob': 85, 'Charlie': 95}. Write code to find the name of the student with the highest score.
• Expected Output:
• Charlie
```

```python
scores = {'Alice': 90, 'Bob': 85, 'Charlie': 95}

top_student = max(scores, key=scores.get)

print(f"Student with highest score: {top_student}")
```
