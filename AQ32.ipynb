```python
33. Given a sentence "apple orange apple banana orange apple", write code to count the occurrences of each word and store the result in a dictionary.
• Expected Output
• {'apple': 3, 'orange': 2, 'banana': 1}
```

```python
sentence = "apple orange apple banana orange apple"

words = sentence.split()

word_counts = {}


for word in words:
  
    word_counts[word] = word_counts.get(word, 0) + 1

print(word_counts)
```
