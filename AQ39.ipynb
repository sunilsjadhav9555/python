```python
1. Common Runtime Errors
These errors happen while the program is running and can be handled using try-except blocks.

ZeroDivisionError: Dividing by zero (e.g., 10 / 0).

IndexError: Accessing a list index that doesn't exist (e.g., list[10]).

KeyError: Accessing a dictionary key that doesn't exist.

TypeError: Doing an operation on the wrong type (e.g., "A" + 5).

ValueError: Correct type but invalid value (e.g., int("hello")).

NameError: Using a variable that hasn't been defined.

FileNotFoundError: Trying to open a file that doesn't exist.

2. Can We Handle Syntax Errors?
Generally, No.

Reason: Syntax errors (like missing a parenthesis or colon) are detected before the program starts running. The code crashes during the "reading" phase,
so the try-except block never even gets a chance to execute.

Exception: You can only catch a SyntaxError if it happens inside dynamic code like eval() or exec().
```
