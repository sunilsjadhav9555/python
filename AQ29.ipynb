```python
30. You are given a list of tuples representing coordinates:
coordinates = [(1, 2), (3, 4), (5, 6)]. Write a loop that unpacks the values in each tuple and prints them as x, y pairs with sum
a. Expected Output
i. x = 1, y = 2 - sm = 3
ii. x = 3, y = 4 - sm = 7
iii. x = 5, y = 6 - sm = 11
```

```python
coordinates = [(1, 2), (3, 4), (5, 6)]

for x, y in coordinates:
    sm = x + y
    print(f"x = {x}, y = {y} - sm = {sm}")
```
