```python
13. Take a username and password as input from the user. 
The username should be compared in a case-insensitive manner, while the password should be compared in a case-sensitive manner. 
The provided username and password should match the following:
a. Username: admin
b. Password: Admin@1234
Print a welcome message if the username and password match otherwise give 3 attempts to the user to provide the username and passwords.
```

```python
CORRECT_USERNAME = "admin"
CORRECT_PASSWORD = "Admin@1234"
MAX_ATTEMPTS = 3

print(" Login System ")

for attempt in range(MAX_ATTEMPTS):
   
    username_input = input("Enter Username: ")
    password_input = input("Enter Password: ")

    if username_input.lower() == CORRECT_USERNAME and password_input == CORRECT_PASSWORD:
        print("\nWelcome! You have successfully logged in.")
        break  
    else:
    
        remaining_attempts = MAX_ATTEMPTS - (attempt + 1)
        
        if remaining_attempts > 0:
            print(f"Incorrect credentials. You have {remaining_attempts} attempt(s) remaining.\n")
        else:
            print("\nAccess Denied. You have exhausted all 3 attempts.")
```
