```python
12. Take a string input from the user and print the string in reverse order using the
following methods:
a. Using a for loop.
b. Using a while loop.
c. Using string slicing.
d. Use reverse method Input: hello
e. Output: olleh
```

```python
text = input("Enter the string:")

rev_str = ""
for char in text:
    rev_str = char + rev_str

print(rev_str)
```

```python
text = input("Enter a string:")
rev_str = ""
i = len(text) - 1

while i >= 0:
    rev_str += text[i]
    i -= 1

print(rev_str)
```

```python
text = input("Enter the string:")
print(text[::-1])
```

```python
text = input("Enter a string:")

char = list(text)
char.reverse()
rev_str = "".join(char)

print(rev_str)
```
