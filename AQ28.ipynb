```python
29. You are developing a system to manage employee records where each employee's information is stored as a tuple with
   (employee_id, name, department, salary). Write a Python program to:
a. Add a new employee record to the list of tuples.
b. Remove an employee record by employee_id.
c. Update the salary for a specific employee_id.
d. Find and print the employee with the highest salary.
employees = [(101, 'Alice', 'HR', 50000), (102, 'Bob', 'IT', 60000), (103, 'Charlie', 'Finance',70000)]
```

```python

employees = [
    (101, 'Alice', 'HR', 50000), 
    (102, 'Bob', 'IT', 60000), 
    (103, 'Charlie', 'Finance', 70000)
]

def add_employee(emp_id, name, dept, salary):
   
    new_record = (emp_id, name, dept, salary)
    employees.append(new_record)
    print(f"Added: {name}")

def remove_employee(emp_id):
    
    for i, emp in enumerate(employees):
        if emp[0] == emp_id:
            removed = employees.pop(i)
            print(f"Removed: {removed[1]} (ID: {emp_id})")
            return
    print("Employee ID not found.")

def update_salary(emp_id, new_salary):
    for i, emp in enumerate(employees):
        if emp[0] == emp_id:
           
            updated_record = (emp[0], emp[1], emp[2], new_salary)
            employees[i] = updated_record
            print(f"Salary updated for {emp[1]} to {new_salary}")
            return
    print("Employee ID not found.")

def find_highest_salary():
    if not employees:
        print("List is empty.")
        return
    highest_earner = max(employees, key=lambda x: x[3])
    
    print(f"Highest Salary Employee: {highest_earner[1]} with {highest_earner[3]}")

print("Original List:", employees)

add_employee(104, "David", "Marketing", 52000)

update_salary(102, 65000)
remove_employee(103)
find_highest_salary()
print("\nFinal List:", employees)
```
