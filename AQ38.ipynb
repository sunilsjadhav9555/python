```python
39. Write a Python program that allows users to search for a recipe by name and saves the result in a new file.
a. Your program should read the data from Recipes.txt, where each recipe contains
the recipe name and its details. Recipes are separated by Recipe Numbers.
b. Prompt the user to enter the name of a recipe they would like to search for.
c. Search through the recipes in recipes.txt to find a matching recipe name. If the recipe is found, retrieve its details.
d. If the recipe is not found, prepare a message saying "Recipe not found."
e. Create a new file named Found_recipe.txt. and If the recipe is found, write the recipe details to this file
```

```python
def search_and_save_recipe():
    # 1. Get user input
    search_term = input("Enter the recipe name to search for: ").strip().lower()
    
    found_recipe_content = []
    is_recording = False
    
    try:
        with open("Recipes.txt", "r") as infile:
            for line in infile:
              
                if line.strip().startswith("Recipe"):
                    
                   
                    if is_recording:
                        break
                    
                 
                    if search_term in line.lower():
                        is_recording = True
                        found_recipe_content.append(line)
                    else:
                        is_recording = False
                
                
                elif is_recording:
                    found_recipe_content.append(line)

        
        with open("Found_recipe.txt", "w") as outfile:
            if found_recipe_content:
                outfile.writelines(found_recipe_content)
                print(f"Success! Recipe details saved to 'Found_recipe.txt'.")
            else:
                outfile.write("Recipe not found.")
                print("Message: Recipe not found.")

    except FileNotFoundError:
        print("Error: 'Recipes.txt' file is missing.")


dummy_recipes = """Recipe 1: Pancakes
Ingredients: Flour, Milk, Eggs, Sugar
Instructions: Mix ingredients and cook on a pan.

Recipe 2: Grilled Cheese Sandwich
Ingredients: Bread, Cheese, Butter
Instructions: Butter bread, place cheese between slices, grill until golden.

Recipe 3: Fruit Salad
Ingredients: Apple, Banana, Orange, Grapes
Instructions: Chop fruits and mix in a bowl.
"""

with open("Recipes.txt", "w") as f:
    f.write(dummy_recipes)


search_and_save_recipe()


print("\n--- Content of Found_recipe.txt ---")
try:
    with open("Found_recipe.txt", "r") as f:
        print(f.read())
except FileNotFoundError:
    pass
```
