```python
32. You are given two sets A = {1, 2, 3, 4} and B = {3, 4, 5, 6}. Write code to find:
• The union of A and B.
• The intersection of A and B.
• The difference between A and B.
• The difference between B and A.
• The difference between A and B. AND The difference between B and A
Note use Operator (|, &, -) as well methods (intersection, union, difference)
```

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(f"Set A: {A}")
print(f"Set B: {B}")
print("-" * 40)


union_op = A | B
union_method = A.union(B)
print(f"1. Union (A | B):  {union_op}")


inter_op = A & B
inter_method = A.intersection(B)
print(f"2. Intersection (A & B): {inter_op}")


diff_ab_op = A - B
diff_ab_method = A.difference(B)
print(f"3. Difference (A - B):  {diff_ab_op}")


diff_ba_op = B - A
diff_ba_method = B.difference(A)
print(f"4. Difference (B - A): {diff_ba_op}")


sym_diff_op = A ^ B
sym_diff_method = A.symmetric_difference(B)
print(f"5. Symmetric Diff (A ^ B):   {sym_diff_op}")
```
