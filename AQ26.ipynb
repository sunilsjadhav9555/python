```python
27. Given a list, write a Python program to create two lists: one with unique values and another with duplicate values.
a. Original_list = [1, 2, 3, 2, 4, 5, 1]
b. Expected Output:
i. unique_list = [3, 4, 5]
ii. duplicate_list = [1, 2]
```

```python
original_list = [1, 2, 3, 2, 4, 5, 1]

unique_list = []
duplicate_list = []

for x in original_list:
    
    if original_list.count(x) == 1:
        unique_list.append(x)
    
    elif original_list.count(x) > 1 and x not in duplicate_list:
        duplicate_list.append(x)

print(f"Unique List:    {unique_list}")
print(f"Duplicate List: {duplicate_list}")
```
