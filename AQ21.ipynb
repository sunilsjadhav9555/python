```python
def fibonacci_series(n):

    a, b = 0, 1
    
   
    if n <= 0:
        print("Please enter a positive integer")
    elif n == 1:
        print(f"Fibonacci sequence upto {n}:")
        print(a)
    else:
        print(f"Fibonacci sequence upto {n}:")
        print(a, end=" ") 
        print(b, end=" ") 
        
        
        for _ in range(2, n):
            c = a + b
            print(c, end=" ")
            
            a = b
            b = c

num_terms = 10
fibonacci_series(num_terms)
```
