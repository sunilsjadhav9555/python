```python
def remove_vowels(text):
    
    vowels = "aeiouAEIOU"
    
    result = "".join([char for char in text if char not in vowels])
    
    return result


input_str = "Hari"
output_str = remove_vowels(input_str)

print(f"Input = {input_str}")
print(f"Output = {output_str}")
```
