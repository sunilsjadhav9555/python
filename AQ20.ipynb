```python
21. Write a Python program to swap the first and last elements of a list or the first and last characters of a string. 
Handle cases where the input has only one element or character. List Input:
a. Input: [1, 2, 3, 4, 5]
b. Expected Output: [5, 2, 3, 4, 1]
```

```python
def swap_list_elements(my_list):
   
    if len(my_list) < 2:
        return my_list
    
    my_list[0], my_list[-1] = my_list[-1], my_list[0]
    
    return my_list


input_list = [1, 2, 3, 4, 5]
output_list = swap_list_elements(input_list)

print(f"Input List:  [1, 2, 3, 4, 5]") 
print(f"Output List: {output_list}")
```
