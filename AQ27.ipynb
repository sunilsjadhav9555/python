```python
28. Given a list of lists, write a Python program to flatten it into a single list containing all the elements of the nested lists.
a. nested_list = [[1, 2], [3, 4], [5, 6]]
b. Expected Output: [1, 2, 3, 4, 5, 6]
```

```python
nested_list = [[1, 2], [3, 4], [5, 6]]

flat_list = [item for sublist in nested_list for item in sublist]

print(f"Original: {nested_list}")
print(f"Flattened: {flat_list}")
```
