```python
def print_diamond_pattern():
    
    rows = 4 

  
    for i in range(1, rows + 1):
     
        star_count = (2 * i) - 1
        
      
        space_count = rows - i
        
  
        line = ("  " * space_count) + ("* " * star_count)
        
        
        print(line.rstrip())

  
    for i in range(rows - 1, 0, -1):
        star_count = (2 * i) - 1
        space_count = rows - i
        
        line = ("  " * space_count) + ("* " * star_count)
        print(line.rstrip())


print_diamond_pattern()
```
