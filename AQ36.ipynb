```python
37. Write a python program to remove the duplicate name from the file.
a. Read the names from List_of_Customers.txt.
b. Removes duplicates while maintaining the original order.
c. Writes the unique names to a new file called Unique_List_Of_Customers.txt.
```

```python
def remove_duplicates_preserve_order(input_file, output_file):
    seen_names = set()
    
    try:
        with open(input_file, 'r') as infile, open(output_file, 'w') as outfile:
            for line in infile:
                name = line.strip()
                
                if not name:
                    continue
                
                if name not in seen_names:
                    outfile.write(name + '\n')
                    seen_names.add(name)
                    
        print(f"Success! Unique names written to '{output_file}'.")
        
    except FileNotFoundError:
        print(f"Error: The file '{input_file}' was not found.")

with open("List_of_Customers.txt", "w") as f:
    f.write("Alice\nBob\nAlice\nCharlie\nBob\nDavid\n")

remove_duplicates_preserve_order("List_of_Customers.txt", "Unique_List_Of_Customers.txt")

print("\n--- Unique List Content ---")
with open("Unique_List_Of_Customers.txt", "r") as f:
    print(f.read())
```
