```python
36. Write a Python program to read and merge the content of two files while organizing the data in a structured way. 
The merged data should be written to a third file with clear section headers.
a. Read the content from both File_Data_1.txt and File_Data_2.txt
b. Merge the content from both files into a new file called organized_data.txt.
c. Add meaningful headers to clearly separate the contents of each file in the new file. For example, 
add a header like "Content from File 1" for the content of data1.txt, and "Content from File 2" for data2.txt
```

```python
def merge_files_with_headers(file1_path, file2_path, output_path):
    try:
     
        with open(output_path, 'w') as outfile:
            
         
            outfile.write("=" * 30 + "\n")
            outfile.write(f"--- CONTENT FROM {file1_path} ---\n")
            outfile.write("=" * 30 + "\n\n")
            
            try:
                with open(file1_path, 'r') as f1:
                    content1 = f1.read()
                  
                    if not content1.endswith('\n'):
                        content1 += '\n'
                    outfile.write(content1)
            except FileNotFoundError:
                outfile.write(f"[Error: {file1_path} was not found.]\n")

        
            outfile.write("\n\n")

           
            outfile.write("=" * 30 + "\n")
            outfile.write(f"--- CONTENT FROM {file2_path} ---\n")
            outfile.write("=" * 30 + "\n\n")
            
            try:
                with open(file2_path, 'r') as f2:
                    content2 = f2.read()
                    outfile.write(content2)
            except FileNotFoundError:
                outfile.write(f"[Error: {file2_path} was not found.]\n")

        print(f"Successfully merged files into '{output_path}'.")

    except Exception as e:
        print(f"An unexpected error occurred: {e}")


with open("File_Data_1.txt", "w") as f:
    f.write("Hello! This is the first line of file 1.\nThis is the second line.")
with open("File_Data_2.txt", "w") as f:
    f.write("Welcome to File 2.\nHere is some more data.")


merge_files_with_headers("File_Data_1.txt", "File_Data_2.txt", "organized_data.txt")

print("\n--- organized_data.txt Content ---")
with open("organized_data.txt", "r") as f:
    print(f.read())
```
