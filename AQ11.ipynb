```python
11. Write a python program to search for a specific word in a list of strings. Use a for
loop with break to exit early once the word is found.
a. 1.Fruit_list = ["apple", "banana", "cherry", "date", "fig"]
b. input word="cherry"
c. Output: Found 'cherry' in the Fruit Bucket!
d. Input word "grapes"
e. Output: 'grapes' not found in the fruit bucket try another fruit.
```

```python
fruit_list = ["apple","banana","cherry","date","fig"]
word = input("Enter the fruit name:")

found = False

for fruit in fruit_list:
    if fruit == word:
        print(f"found {word} in the fruit bucket")
        found = True
        break

if not found:
    print(f"{word} not found in the fruit bucket, try another fruit.")
```
