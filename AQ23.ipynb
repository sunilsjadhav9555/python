```python
24. Given a list, split it into two separate lists from the halfway point. If the list has an odd number of elements, 
include the middle element in the first list.

a. original_list = [1, 2, 3, 4, 5, 6, 7]

b. Expected Output:

i. lst_first_half = [1, 2, 3, 4]

ii. lst_second_half = [5, 6, 7]
```

```python
def split_list_halfway(my_list):
    length = len(my_list)
    
    split_idx = (length + 1) // 2

    first_half = my_list[:split_idx]
    second_half = my_list[split_idx:]
    
    return first_half, second_half


original_list = [1, 2, 3, 4, 5, 6, 7]
lst_first, lst_second = split_list_halfway(original_list)

print(f"Original List: {original_list}")
print(f"First Half:  {lst_first}")
print(f"Second Half: {lst_second}")
```
