```python
14. Given a string containing both alphabetic characters and numeric digits, 
convert the string into a sorted sequence with alphabetic characters followed by numeric digits. The characters should be sorted in alphabetical order,
and the digits should be sorted in numerical order. Example:
a. Input: Al249ntxrt7879
b. Output: Alnorttx2477899
```

```python
def sort_alpha_numeric(input_string):
    
    alphas = []
    digits = []

 
    for char in input_string:
        if char.isalpha():
            alphas.append(char)
        elif char.isdigit():
            digits.append(char)

  
    alphas.sort() 
    digits.sort()

    
    sorted_alphas = "".join(alphas)
    sorted_digits = "".join(digits)

    return sorted_alphas + sorted_digits

input_str = "Al249ntxrt7879"
output_str = sort_alpha_numeric(input_str)

print(f"Input:  {input_str}")
print(f"Output: {output_str}")
```
