```python
26. Given a list, write a Python program to create two lists: one with unique values and another with duplicate values.
a. Original_list = [1, 2, 3, 2, 4, 5, 1]
b. Expected Output:
i. unique_list = [3, 4, 5]
ii. duplicate_list = [1, 2,2,1]
Note: Use List comprehension.
```

```python
original_list = [1, 2, 3, 2, 4, 5, 1]

unique_list = [x for x in original_list if original_list.count(x) == 1]

duplicate_list = [x for x in original_list if original_list.count(x) > 1]

print(f"Original List:  {original_list}")
print(f"Unique List:    {unique_list}")
print(f"Duplicate List: {duplicate_list}")
```
