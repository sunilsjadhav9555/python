```python
def check_anagram_number(num1, num2):
    
    str1 = str(num1)
    str2 = str(num2)
    
    if len(str1) != len(str2):
        return False

    
    if sorted(str1) == sorted(str2):
        return True
    else:
        return False

n1 = 12345
n2 = 54321
if check_anagram_number(n1, n2):
    print(f"{n1} and {n2} ARE Anagrams.")
else:
    print(f"{n1} and {n2} are NOT Anagrams.")


n3 = 123
n4 = 122
if check_anagram_number(n3, n4):
    print(f"{n3} and {n4} ARE Anagrams.")
else:
    print(f"{n3} and {n4} are NOT Anagrams.")
```
