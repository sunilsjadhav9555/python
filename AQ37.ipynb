```python
38. Write a Python program to process the grades in Grades.txt, calculate the average grade, and identify students who scored above average.
a. Your program should read the data from Grades.txt. Each line contains a student name and their grade, separated by a comma.
b. Extract the grades from the file and compute the average grade of all students. Create a list of students whose grades are above the calculated average.
c. Create a new file named Above_average_students.txt
d. Write the average grade to this file, followed by the names of the students who scored above average, each on a new line.
```

```python
def process_grades(input_file, output_file):
    students = []
    
    try:
     
        with open(input_file, 'r') as f:
            for line in f:
               
                parts = line.strip().split(',')
                
                if len(parts) == 2:
                    name = parts[0].strip()
                    try:
                        grade = float(parts[1].strip())
                        students.append((name, grade))
                    except ValueError:
                        print(f"Skipping invalid grade for {name}")

  
        if not students:
            print("No valid student data found.")
            return

        total_score = sum(student[1] for student in students)
        average_grade = total_score / len(students)
        
       
        above_avg_students = [student[0] for student in students if student[1] > average_grade]

  
        with open(output_file, 'w') as f:
            f.write(f"Average Grade: {average_grade:.2f}\n")
            f.write("-" * 20 + "\n")
            f.write("Students Above Average:\n")
            
            for name in above_avg_students:
                f.write(name + '\n')

        print(f"Success! Average ({average_grade:.2f}) and {len(above_avg_students)} names written to '{output_file}'.")

    except FileNotFoundError:
        print(f"Error: The file '{input_file}' was not found.")

dummy_data = """Alice, 85
Bob, 78
Charlie, 92
David, 60
Eve, 88
Frank, 75"""

with open("Grades.txt", "w") as f:
    f.write(dummy_data)
process_grades("Grades.txt", "Above_average_students.txt")
print("\n--- Above_average_students.txt Content ---")
with open("Above_average_students.txt", "r") as f:
    print(f.read())
```
