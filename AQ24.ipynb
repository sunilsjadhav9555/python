```python
25. Given a list, rotate (or shuffle) its elements by nth positions to the left. After the shift, 
elements that are moved out from the beginning of the list should appear at the end.
a. Original_list = ['a', 'b', 'c', 'd', 'e']
b. n = 2
c. Expected Output - Shuffled_list = ['c', 'd', 'e', 'a', 'b']
```

```python
def rotate_list_left(my_list, n):
   
    if len(my_list) == 0:
        return my_list
    n = n % len(my_list)
    rotated_list = my_list[n:] + my_list[:n]
    
    return rotated_list

original_list = ['a', 'b', 'c', 'd', 'e']
n = 2

shuffled_list = rotate_list_left(original_list, n)

print(f"Original List: {original_list}")
print(f"n = {n}")
print(f"Shuffled List: {shuffled_list}")
```
